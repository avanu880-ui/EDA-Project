```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sbn

df=pd.read_csv(r"D:\python programs\WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
dimension=df.shape
print("Number of columns:",dimension[1])
print("Number of rows:",dimension[0])
df.head()
```

```python
#column mames
print("Column Name:",df.columns)
df.dtypes
```

```python
#TRANSFORM TOTAL CHARGES INTO NUMERICAL FIELD
df["TotalCharges"]=df["TotalCharges"].replace(" ","0")
df["TotalCharges"] = df["TotalCharges"].astype("float")
df.dtypes
```

```python
#Transform totalcharges into numerical field (secound way)
df.dtypes
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"],errors='coerce')
df.dtypes
```

```python
print("Total Null value column wise\n\n",df.isnull().sum())
print("\nTotal Null value in all column\n\n",df.isnull().sum().sum())
```

```python
#replace Nan with zero
df["TotalCharges"]=df["TotalCharges"].replace(np.nan, 0)
print("Total Null value column wise\n\n",df.isnull().sum())
print("\nTotal Null value in all column\n\n",df.isnull().sum().sum())
```

```python
#Number of churn in our data
plt.figure(figsize=(3,4))
ax=sbn.countplot(x = "Churn",data=df)
#print(ax.containers)
ax.bar_label(ax.containers[0])
plt.title("count with churn data")
plt.show()
```

```python
#prepare pie chart
churn_counts = df["Churn"].value_counts()
churn_counts
plt.figure(figsize=(3,3))
plt.pie(
    churn_counts,
    labels= churn_counts.index,
    startangle = 90,
    autopct = "%1.1f%%"
    



)
plt.title("PERCENTAGE DISTRIBUTION OF CHART")
plt.show()
```

```python

# Create a cross-tab of Gender vs Churn
gender_churn = df.groupby(["gender", "Churn"]).size().unstack(fill_value=0)

# Plot stacked bar chart
gender_churn.plot(
    kind="bar",
    stacked=True,
    figsize=(6, 4),
    edgecolor="black"
)

# Add title and labels
plt.title("Gender-wise Customer Churn Distribution", fontsize=12, fontweight="bold")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.legend(title="Churn Status", loc="upper right")

# Add percentage labels inside bars
for i, gender in enumerate(gender_churn.index):
    total = gender_churn.loc[gender].sum()
    bottom = 0
    for churn_status, count in gender_churn.loc[gender].items():
        plt.text(
            i, 
            bottom + count / 2, 
            f"{count} ({count/total:.1%})", 
            ha="center", 
            va="center", 
            color="white", 
            fontweight="bold",
            fontsize=9
        )
        bottom += count

plt.tight_layout()
plt.show()
```
